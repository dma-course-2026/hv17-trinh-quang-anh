```python
#Bài 50
def tim_so():
    for i in range(100, 1000):
        hangtram = i//100
        hangchuc = (i//10) %10
        hangdonvi = i%10
        tong = hangtram+hangchuc+hangdonvi
        tich = hangtram*hangchuc*hangdonvi
        if tong == tich:
            print(hangtram, hangchuc, hangdonvi)
tim_so()
```

```python
#Bài 51
def tien():
    for so_500 in range(0,4):
        for so_200 in range(0,9):
            for so_100 in range(0,18):
                tong_tien = so_500 * 500 +so_200 * 200 +so_100 * 100
                if tong_tien == 1700:
                    print("so to 500:", so_500, end=" ")
                    print("; so to 200:", so_200, end=" ")
                    print("; so to 100:", so_100)
tien()
```

```python
#Bài 52
def doi_tien():
    n = int(input("so tien can doi n ="))
    t1 = int(input("loai tien thu 1:"))
    t2 = int(input("loai tien thu 2:"))
    t3 = int(input("loai tien thu 3:"))
    print(f"Loai {t1} Loai {t2} Loai {t3}")
    for s1 in range(0, n // t1 +1):
        for s2 in range(0, n // t2 +1):
            for s3 in range(0, n // t3 +1):
                if s1 *t1 +s2 *t2 +s3 *t3 == n:
                    print(s1, s2, s3)
doi_tien()
```

```python
#Bài 55
def giaithua(n):
    ketqua = 1
    for i in range(1, n+1):
        ketqua = ketqua*i
    return ketqua
n = int(input("nhap n: "))
print(giaithua(n))
```

```python
#Bài 56
def giai_thua(x):
    tich = 1
    for i in range(1, x+1):
        tich = tich *i
    return tich
def to_hop(k, n):
    return giai_thua(n) / (giai_thua(k)*giai_thua(n -k))
n = int(input("Nhap n = "))
k = int(input("Nhap k = "))
kq = to_hop(k, n)
print("C(", n,",", k,") =", kq)
```

```python
#Bài 57
def fibonacci(n):
    n = int(input("Nhap n: "))
    a = 0
    b = 1
    for i in range(n):
        print(a, end=" ")
        tieptheo = a +]b
        a = b
        b = tieptheo
    print(f"{n} số đầu tiên của dãy số fibonacci: ")
fibonacci(n)
```

```python
#Bài 58
def uscln(a, b):
    uscln = 1
    for i in range(1, min(a, b) +1):
        if a%i == 0 and b%i == 0:
            uscln = i
    return uscln


def bscnn(a, b):
    bscnn = (a * b) // uscln(a, b)
    return bscnn


a = int(input("so nguyen duong a: "))
b = int(input("so nguyen duong b: "))

ucln = uscln(a, b)
bcnn = bscnn(a, b)

print("uoc chung lon nhat cua", a, "va", b, "la:", ucln)
print("boi chung lon nhat cua", a, "va", b, "la:", bcnn)
```
