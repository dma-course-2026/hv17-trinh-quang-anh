```python
#Bài 4
def noi_chuoi():
    #khi đã nhập input thì nó mặc định là 1 chuỗi hay 1 str rồi
    n = input("nhập n: ")
    m = input("nhập m: ")
    c = n + " " + m
    #chỗ " " là dấu cách
    print(type(c))
    print(f"{c}")
noi_chuoi()
```

```python
#Bài 5
def chon_chuoi_dai():
    n = input("chuỗi n: ")
    m = input("chuỗi m: ")
    if len(n) > len(m):
        #len ở đây dùng để so sánh có bao nhiêu kí tự trong 2 chuỗi
        print(type(n))
        print(f"{n}")
    elif len(n) < len(m):
        print(type(m))
        print(f"{m}")
    else:
        print("hai chuỗi có độ dài bằng nhau")
chon_chuoi_dai()
```
