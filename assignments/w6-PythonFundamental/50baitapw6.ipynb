```python
# Bai 2
def hai_so_nguyen():
    a = int(input("Số nguyên thứ nhất là: "))
    b = int(input("Số nguyên thứ hai là:"))
    tong = a + b
    hieu = a - b
    tich = a * b
    thuong = a / b
    print ("Tổng 2 số: ", tong)
    print ("Hiệu 2 số: ", hieu)
    print ("Tích 2 số: ", tich)
    print ("Thương 2 số: ", thuong)
    
hai_so_nguyen()
```

```python
#Bai 5
def thi_dua():
    a = int(input("Không đeo thẻ: "))
    b = int(input("Nói chuyện trong lớp: "))
    c = int(input("Đi học muộn: "))
    t = int(input("Trường hợp không đeo thẻ: "))
    n = int(input("Trường hợp nói chuyện riêng: "))
    m = int(input("Trường hợp đi học muộn: "))
    tru_diem = t*a + n*b + m*c
    print ("Tổng điểm thi đua bị trừ: ", tru_diem)
    
thi_dua()
```

```python
#Bai 6
def khu_vuon():
    a = int(input("Chiều rộng khu vườn: "))
    b = int(input("Chiều dài khu vườn: "))
    con_lai = a*b - (3.14*(a/2)**2)
    print ("Diện tích còn lại của khu vườn: ", con_lai)

khu_vuon()
```

```python
#Bai 7
def diem_thi():
    toan = int(input("Điểm môn toán: "))
    van = int(input("Điểm môn văn: "))
    anh = int(input("Điểm môn anh: "))
    trung_binh = (toan + van + anh)/3
    if trung_binh >= 5:
        print ("Qua mon")
    else:
        print ("Truot mon")

diem_thi()
```

```python
#Bai 8
def ki_tu():
    c = input("kí tự: ")
    if 'A' <= c <= 'Z' or 'a' <= c <= 'z':
        print ("la ki tu alphabet: ")
    else:
        print ("khong phai la ki tu alphabet: ")

ki_tu()
```
