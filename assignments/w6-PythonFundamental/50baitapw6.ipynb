```python
#Bai 14
def hinh_tam_giac():
    a = int(input("cạnh a: "))
    b = int(input("cạnh b: "))
    c = int(input("cạnh c: "))
    if (a > 0 and b > 0 and c > 0) and (a + b > c) and (a + c > b) and (b + c > a):
        chu_vi = a + b + c
        p = chu_vi / 2
        dien_tich = p * (p - a) * (p - b) * (p - c)
        print("chu vi tam giác: ", chu_vi)
        print("diện tích tam giác: ", dien_tich)
    else:
        print(f"{a:g}, {b:g}, {c:g} không phải là 3 cạnh của tam giác")

hinh_tam_giac()
```

```python
#Bài 20
def giai_phuong_trinh_bac_2():
    a = int(input("Nhập a: "))
    b = int(input("Nhập b: "))
    c = int(input("Nhập c: "))
    if a == 0:
        if b == 0:
            if c == 0:
                print("Phương trình có vô số nghiệm")
            else:
                print("Phương trình vô nghiệm")
        else:
            x = -c / b
            print("Phương trình có một nghiệm: x = ", x)
    else:
        delta = b**2 - 4 * a * c
        
        if delta < 0:
            print("Phương trình vô nghiệm")
        elif delta == 0:
            x = -b / (2 * a)
            print("Phương trình có nghiệm kép: x1 = x2 = ", x)
        else:
            x1 = (-b + (delta)) / (2 * a)
            x2 = (-b - (delta)) / (2 * a)
            print("Phương trình có 2 nghiệm phân biệt:")
            print("x1 = ", x1)
            print("x2 = ", x2)

giai_phuong_trinh_bac_2()
```

```python
#Bài 33
def diem_thuoc_duong_tron():
    R = int(input("Nhập bán kính R: "))
    x0 = int(input("Hoành độ tâm đường tròn: "))
    y0 = int(input("Tung độ tâm đường tròn: "))
    xa = int(input("Hoành độ điểm A: "))
    ya = int(input("Tung độ điểm A: "))
    d = ((xa-x0)**2 + (ya-y0)**2)**0.5
    if d<R:
        print ("Diem A nam trong duong tron")
    elif d==R:
        print ("Diem A nam tren duong tron")
    else:
        print ("Diem A nam ngoai duong tron")

diem_thuoc_duong_tron()
```
