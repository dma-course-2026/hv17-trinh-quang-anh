```python
#Bai 1
def tim_cac_so():
    for i in range(2000, 3201):
        if i % 7 == 0 and i % 5 != 0:
            print(i, end=", ")

tim_cac_so()
```

```python
#Bai 2
def tinh_giai_thua(n):
    ket_qua = 1
    for i in range(1, n + 1):
        ket_qua = ket_qua * i
    return ket_qua

so = 8
print(tinh_giai_thua(so))
```

```python
#Bai 3
def tao_dictionary(n):
    ket_qua = {}
    for i in range(1, n + 1):
        ket_qua[i] = i * i
    return ket_qua

n = int(input("Nhập số nguyên n: "))
print(tao_dictionary(n))
```

```python
#Bai 4
def tao_danh_sach_va_tuple():
    chuoi = input("Nhập một chuỗi số, phân tách bằng dấu phẩy: ")
    
    danh_sach = chuoi.split(",")
    
    tup = tuple(danh_sach)
    
    print(danh_sach)
    print(tup)

tao_danh_sach_va_tuple()
```

```python
#Bai 5
class IOString:
    def __init__(self):
        self.s = ""

    def getString(self):
        self.s = input("Nhập một chuỗi: ")

    def printString(self):
        print(self.s.upper())

def kiem_tra_class():
    obj = IOString()
    
    obj.getString()
    
    obj.printString()

kiem_tra_class()
```

```python
#Bai 9
from datetime import datetime
def tinh_tuoi():
    chuoi_ns = input("Nhập ngày sinh (y/m/d): ")
    #nhập ngày tháng năm sinh theo ví dụ: 2008/10/10
    
    ngay_sinh = datetime.strptime(chuoi_ns, "%Y/%m/%d")
    
    ngay_hien_tai = datetime.now()
    
    tuoi = ngay_hien_tai.year - ngay_sinh.year
    
    if (ngay_hien_tai.month, ngay_hien_tai.day) < (ngay_sinh.month, ngay_sinh.day):
        tuoi -= 1
        
    print(f"Tuổi của bạn là: {tuoi}")

tinh_tuoi()
```

```python
#Bai 10
def tinh_tien_luong():
    so_gio = float(input("Nhập số giờ làm mỗi tuần: "))
    thu_lao_chuan = float(input("Nhập thù lao trên mỗi giờ làm tiêu chuẩn: "))
    
    gio_tieu_chuan = 44
    
    if so_gio <= gio_tieu_chuan:
        thuc_linh = so_gio * thu_lao_chuan
    else:
        gio_vuot = so_gio - gio_tieu_chuan
        thuc_linh = (gio_tieu_chuan * thu_lao_chuan) + (gio_vuot * thu_lao_chuan * 1.5)
        
    print(f"Số tiền thực lĩnh của nhân viên là: {thuc_linh}")

tinh_tien_luong()
```
